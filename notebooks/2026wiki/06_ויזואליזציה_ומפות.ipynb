## חלק א: עקרונות ויזואליזציה טובה
### מדוע ויזואליזציה חשובה?
**"A picture is worth a thousand words"** – אבל רק אם הוא מדויק ומכבד את הנתונים.
### עקרונות Edward Tufte:
| עיקרון | תיאור | דוגמה |
|--------|-------|--------|
| **Data-Ink Ratio** | מקסמו "דיו של נתונים" – הסירו קישוטים | הסירו gridlines מיותרות |
| **Chartjunk** | הימנעו מקישוטים ללא תוכן | ללא 3D, ללא הצללות |
| **Lie Factor** | ייצוג יחסי נאמן לנתונים | ציר Y שמתחיל ב-0 |
| **Small Multiples** | השוואות ע"י ריבוי גרפים זהים | פאנל גרפים לפי שנה |
### סוגי גרפים וייעודם:
| גרף | מתי להשתמש |
|-----|------------|
| **עמודות (Bar)** | השוואת קטגוריות |
| **קו (Line)** | מגמות לאורך זמן |
| **פיזור (Scatter)** | קשר בין שני משתנים |
| **היסטוגרמה** | התפלגות ערכים |
| **חום (Heatmap)** | מטריצות ומתאמים |
| **מפה (Map)** | נתונים גיאוגרפיים |
📖 לקריאה: Tufte, E. (1983). *The Visual Display of Quantitative Information*. Graphics Press.

## חלק ב: גרפים סטטיים – Matplotlib ו-Seaborn
### מתי גרפים סטטיים?
- **פרסום** – מאמרים, כנסים
- **רפורטים** – PDF, Word
- **פשטות** – גרפים בסיסיים שלא דורשים אינטראקציה

```python
# ============================================================
# גרפים סטטיים – ניתוח אתרים ארכיאולוגיים
# ============================================================

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('ניתוח אתרים ארכיאולוגיים בישראל', fontsize=16, fontweight='bold', y=0.98)

REGION_COLORS = {'צפון': '#e74c3c', 'מרכז': '#2980b9', 'דרום': '#27ae60'}

# ── גרף 1: שכבות לפי אתר ──
ax1 = axes[0, 0]
df_sorted = df_sites.sort_values('שכבות', ascending=True)
colors1 = [REGION_COLORS[r] for r in df_sorted['אזור']]
bars = ax1.barh(df_sorted['שם'], df_sorted['שכבות'], color=colors1, alpha=0.85)
ax1.set_xlabel('מספר שכבות', fontsize=11)
ax1.set_title('שכבות התיישבות לפי אתר', fontsize=12, fontweight='bold')
for bar, val in zip(bars, df_sorted['שכבות']):
    ax1.text(bar.get_width() + 0.2, bar.get_y() + bar.get_height()/2,
             str(val), va='center', fontsize=9)

# ── גרף 2: התפלגות אזורית ──
ax2 = axes[0, 1]
region_counts = df_sites['אזור'].value_counts()
wedge_colors = [REGION_COLORS[r] for r in region_counts.index]
wedges, texts, autotexts = ax2.pie(
    region_counts, labels=region_counts.index,
    colors=wedge_colors, autopct='%1.0f%%',
    startangle=90, textprops={'fontsize': 11}
)
ax2.set_title('חלוקה אזורית', fontsize=12, fontweight='bold')

# ── גרף 3: שכבות לפי חשיבות ──
ax3 = axes[1, 0]
scatter_colors = [REGION_COLORS[r] for r in df_sites['אזור']]
sc = ax3.scatter(df_sites['חשיבות'], df_sites['שכבות'],
                 c=scatter_colors, s=df_sites['חשיבות'] * 20,
                 alpha=0.8, edgecolors='white', linewidth=1.5)
for _, row in df_sites.iterrows():
    ax3.annotate(row['שם'], (row['חשיבות'], row['שכבות']),
                 textcoords='offset points', xytext=(5, 3), fontsize=8, color='#555')
ax3.set_xlabel('חשיבות היסטורית', fontsize=11)
ax3.set_ylabel('מספר שכבות', fontsize=11)
ax3.set_title('שכבות לעומת חשיבות', fontsize=12, fontweight='bold')

# ── גרף 4: boxplot תקופות ──
ax4 = axes[1, 1]
region_data = [df_sites[df_sites['אזור'] == r]['שכבות'].tolist()
               for r in ['צפון', 'מרכז', 'דרום']]
bp = ax4.boxplot(region_data, labels=['צפון', 'מרכז', 'דרום'],
                 patch_artist=True)
for patch, color in zip(bp['boxes'], ['#e74c3c', '#2980b9', '#27ae60']):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)
ax4.set_ylabel('מספר שכבות', fontsize=11)
ax4.set_title('התפלגות שכבות לפי אזור', fontsize=12, fontweight='bold')

# מקרא
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor=c, label=r) for r, c in REGION_COLORS.items()]
fig.legend(handles=legend_elements, loc='lower center', ncol=3,
           fontsize=10, title='אזור', bbox_to_anchor=(0.5, -0.02))

plt.tight_layout()
plt.savefig('01_static_charts.png', dpi=150, bbox_inches='tight')
plt.show()
print("💾 הגרף נשמר: 01_static_charts.png")
```

## חלק ג: גרפים אינטראקטיביים עם Plotly
### יתרונות Plotly:
- **Hover**: מידע מפורט בעת ריחוף
- **Zoom**: זום ידני
- **Click**: הדגשה/הסתרה
- **Export**: PNG, HTML, SVG
### Plotly Express לעומת Plotly Graph Objects:
- `px` (Express) = ממשק פשוט, קוד קצר
- `go` (Graph Objects) = שליטה מלאה, קוד ארוך יותר

```python
# ============================================================
# גרפים אינטראקטיביים – Plotly
# ============================================================

print("📊 יוצר גרפים אינטראקטיביים...")

# ── גרף פיזור אינטראקטיבי ──
fig_scatter = px.scatter(
    df_sites,
    x='חשיבות',
    y='שכבות',
    size='שכבות',
    color='אזור',
    hover_name='שם',
    hover_data={'תקופה': True, 'שכבות': True, 'חשיבות': True},
    color_discrete_map={'צפון': '#e74c3c', 'מרכז': '#2980b9', 'דרום': '#27ae60'},
    title='אתרים ארכיאולוגיים – שכבות לעומת חשיבות',
    labels={'חשיבות': 'חשיבות היסטורית', 'שכבות': 'מספר שכבות'},
    size_max=40
)
fig_scatter.update_layout(
    font_family="Arial",
    title_font_size=16,
    legend_title_text='אזור',
    height=450
)
fig_scatter.write_html('02_scatter_interactive.html')
fig_scatter.show()
print("✅ נשמר: 02_scatter_interactive.html")

# ── גרף עמודות מקובצות ──
fig_bar = px.bar(
    df_sites,
    x='שם',
    y='שכבות',
    color='אזור',
    color_discrete_map={'צפון': '#e74c3c', 'מרכז': '#2980b9', 'דרום': '#27ae60'},
    title='מספר שכבות לכל אתר',
    labels={'שם': 'שם האתר', 'שכבות': 'מספר שכבות'},
    hover_data=['תקופה', 'חשיבות']
)
fig_bar.update_layout(xaxis_tickangle=-45, height=450)
fig_bar.write_html('03_bar_interactive.html')
fig_bar.show()
print("✅ נשמר: 03_bar_interactive.html")
```

## חלק ד: מפות גיאוגרפיות עם Folium
### מהו Folium?
**Folium** = ספריית Python ליצירת מפות אינטראקטיביות המבוססות על **Leaflet.js**.
### סוגי מפות:
| סוג | שימוש |
|-----|-------|
| **Marker** | נקודות בודדות (אתרים, ממצאים) |
| **CircleMarker** | עיגולים בגדלים שונים (כמות) |
| **Heatmap** | צפיפות נקודות |
| **Choropleth** | צביעת אזורים לפי ערך |
| **GeoJSON** | שכבות ישויות גיאוגרפיות |
### GIS בארכיאולוגיה:
**GIS** (Geographic Information System) = מערכת מידע גיאוגרפי.
- ניתוח מרחבי של ממצאים
- זיהוי דפוסי התיישבות
- שחזור נופי עבר (Landscape Archaeology)
- ניתוח חזות (Viewshed Analysis)

```python
# ============================================================
# מפות גיאוגרפיות – Folium
# ============================================================

print("🗺️  יוצר מפות גיאוגרפיות...")

# ── מפה 1: אתרים ארכיאולוגיים ──
m1 = folium.Map(
    location=[31.5, 35.0],   # מרכז ישראל
    zoom_start=8,
    tiles='OpenStreetMap'
)

# הוספת כותרת
title_html = '''
<div style="position: fixed; top: 10px; left: 50%; transform: translateX(-50%);
            background: white; padding: 8px 16px; border-radius: 8px;
            border: 2px solid #2980b9; font-size: 16px; font-weight: bold; z-index: 1000;">
    אתרים ארכיאולוגיים בישראל
</div>
'''
m1.get_root().html.add_child(folium.Element(title_html))

REGION_COLORS_FOLIUM = {'צפון': 'red', 'מרכז': 'blue', 'דרום': 'green'}

for _, site in df_sites.iterrows():
    # גודל העיגול = חשיבות
    site_name = site['שם']
    site_period = site['תקופה']
    site_layers = site['שכבות']
    site_importance = site['חשיבות']
    site_region = site['אזור']
    
    folium.CircleMarker(
        location=[site['lat'], site['lon']],
        radius=site['חשיבות'] * 2,
        color=REGION_COLORS_FOLIUM.get(site['אזור'], 'gray'),
        fill=True,
        fill_opacity=0.7,
        popup=folium.Popup(
            '<div dir="rtl" style="text-align:right; font-family:Arial;">'
            '<b>' + site_name + '</b><br>'
            '<b>תקופה:</b> ' + site_period + '<br>'
            '<b>שכבות:</b> ' + str(site_layers) + '<br>'
            '<b>חשיבות:</b> ' + str(site_importance) + '/10<br>'
            '<b>אזור:</b> ' + site_region +
            '</div>',
            max_width=200
        ),
        tooltip=site_name
    ).add_to(m1)

# מקרא
legend_html = '''
<div style="position: fixed; bottom: 30px; right: 10px;
            background: white; padding: 10px; border-radius: 8px;
            border: 2px solid gray; font-family: Arial; font-size: 13px; z-index: 1000;">
    <b>מקרא – אזורים:</b><br>
    צפון &nbsp; מרכז &nbsp; דרום<br>
    <i>גודל עיגול = חשיבות</i>
</div>
'''
m1.get_root().html.add_child(folium.Element(legend_html))

m1.save('04_archaeology_map.html')
print("✅ מפה נשמרה: 04_archaeology_map.html")

# הצגה ב-Colab
display(IFrame('04_archaeology_map.html', width='100%', height=550))
```

```python
# ============================================================
# מפת צפיפות (Heatmap) – Folium
# ============================================================

try:
    from folium.plugins import HeatMap
    
    # מפה 2: Heatmap של אתרים לפי חשיבות
    m2 = folium.Map(location=[31.5, 35.0], zoom_start=8, tiles='CartoDB positron')
    
    # נתונים: [lat, lon, weight]
    heat_data = [[row['lat'], row['lon'], row['חשיבות']] for _, row in df_sites.iterrows()]
    
    HeatMap(
        heat_data,
        min_opacity=0.3,
        max_zoom=18,
        radius=30,
        blur=20,
        gradient={0.4: 'blue', 0.65: 'lime', 1: 'red'}
    ).add_to(m2)
    
    m2.save('05_heatmap.html')
    print("✅ Heatmap נשמר: 05_heatmap.html")
    display(IFrame('05_heatmap.html', width='100%', height=500))

except ImportError:
    print("⚠️ folium.plugins לא זמין – מדלג על Heatmap")
    print("   pip install folium[plugins]")
```
